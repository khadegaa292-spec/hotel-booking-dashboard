```python
import pandas as pd
```

```python
df=pd.read_csv("hotel_booking.csv")
```

```python
df=df.drop(columns=['name','email','phone-number','credit_card'],errors='ignore')
```

```python
df=df.drop_duplicates()
print(df.shape)
```

```python
df.duplicated().sum()
print(len(df))
```

```python
df['booking_via_agent']=df['agent'].notnull().astype(int)
```

```python
df['booking_via_company']=df['company'].notnull().astype(int)
```

```python
df['agent']=df['agent'].fillna(0)

df['company']=df['company'].fillna(0)

df['country']=df['country'].fillna('Unknown')

df['children']=df['children'].fillna(0).astype(int)
```

```python
df['meal']=df['meal'].replace('undefined','sc')
```

```python
print(df.isnull().sum().sum())
```

```python
no_guests = df['adults'] + df['children'] + df['babies'] == 0
no_nights = df['stays_in_weekend_nights'] + df['stays_in_week_nights'] == 0
bad_adr = df['adr'] <= 0

print(f"Bookings with 0 guests: {no_guests.sum()}")
print(f"Bookings with 0 nights: {no_nights.sum()}")
print(f"Bookings with bad ADR (<= 0): {bad_adr.sum()}")
print(f"Total invalid rows: {(no_guests | no_nights | bad_adr).sum()}")

df[no_guests | no_nights | bad_adr].head()
```

```python
no_guests = df['adults'] + df['children'] + df['babies'] == 0
no_nights = df['stays_in_weekend_nights'] + df['stays_in_week_nights'] == 0
bad_adr = (df['adr'] <= 0) | (df['adr'] > 1000)

print(f"(0 guests): {no_guests.sum()}")
print(f"(0 nights): {no_nights.sum()}")
print(f"(bad ADR): {bad_adr.sum()}")
```

```python
df['total_guests'] = df['adults'] + df['children'] + df['babies']
df['total_nights'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df['revenue'] = df['adr'] * df['total_nights']

print(df[['total_guests', 'total_nights', 'revenue']].head())
```

```python
df['arrival_date'] = pd.to_datetime(
    df['arrival_date_year'].astype(str) + '-' +
    df['arrival_date_month'].astype(str) + '-' +
    df['arrival_date_day_of_month'].astype(str))
```

```python
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 
               'July', 'August', 'September', 'October', 'November', 'December']
df['arrival_date_month'] = pd.Categorical(df['arrival_date_month'], categories=month_order, ordered=True)
```

```python
df = df[(df['total_guests'] > 0) & (df['total_nights'] > 0)]
```

```python
df['is_free_or_invalid'] = (df['adr'] <= 0) | (df['adr'] > 1000)
```

```python
df.shape
```

```python
df_rev = df[df['is_free_or_invalid'] == False]

print("Total Bookings:", len(df))
print("Cancellation Rate %:", round(df['is_canceled'].mean() * 100, 1))
print("Avg ADR:", round(df_rev['adr'].mean(), 2))
print("Total Revenue (not canceled):", round(df_rev[df_rev['is_canceled'] == 0]['revenue'].sum()))
print("Avg Lead Time:", round(df['lead_time'].mean(), 1))
print("Avg Nights:", round(df['total_nights'].mean(), 2))
```
